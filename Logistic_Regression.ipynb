              #Heart Disease Dataset — Column Descriptions
              

1. age

Age of the patient (in years)

2. sex

Gender of the patient

0 = Female

1 = Male

3. cp (Chest Pain Type)

Chest pain experienced by the patient:

0 = Typical angina

1 = Atypical angina

2 = Non-anginal pain

3 = Asymptomatic

Higher values mean less obvious chest-pain symptoms.

4. trestbps

Resting blood pressure (in mm Hg)

Measured during hospital admission

5. chol

Serum cholesterol (mg/dl)

High values indicate risk of heart disease.

6. fbs (Fasting Blood Sugar)

1 = FBS > 120 mg/dl

0 = FBS ≤ 120 mg/dl

High fasting glucose indicates diabetes risk.

7. restecg (Resting ECG Results)

0 = Normal

1 = ST-T wave abnormality

2 = Left ventricular hypertrophy

8. thalach

Maximum heart rate achieved

Lower values → weaker heart performance.

9. exang

Exercise-induced angina

1 = Yes (pain during exercise)

0 = No

10. oldpeak

ST depression induced by exercise relative to rest

A measure of heart stress

Larger values → more severe heart issues.

11. slope

Slope of the peak exercise ST segment:

0 = Upsloping

1 = Flat

2 = Downsloping

Downsloping (2) is the most concerning.

12. ca

Number of major vessels colored by fluoroscopy

Range: 0–4

Higher values = higher severity.

13. thal

Thalassemia blood disorder test:

0 = Unknown

1 = Normal

2 = Fixed defect

3 = Reversible defect

14. target

Heart disease diagnosis

1 = Heart disease present

0 = No heart disease

This is the prediction label used in ML.

#Split the data into features(x)and target(y) label sets

```python
x=pd.DataFrame(dataset.iloc[:,:-1])
y=pd.DataFrame(dataset.iloc[:,-1])
```

#Split the data into a training amd test set

```python
#import module to split dataset
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

#Create and Train the model

```python
#import module for fitting
from sklearn.linear_model import LogisticRegression

#creat instance(i.e object) of LogisticRegression
model=LogisticRegression()
model.fit(x_train,y_train)
```

#Predecting the test set results

```python
y_pred=model.predict(x_test)
```

```python
len(y_pred)
```

```python
y_test
```

#Calculating the accuracy

```python
print(f"Accuracy: {model.score(x_test, y_test)*100}")
```

```python
print(f"Accuracy: {model.score(x_train, y_train)*100}")
```

#Evaluate model using confusion matrix

```python
#Confusion matrix
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_train,model.predict(x_train))
print(cm)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_train,model.predict(x_train)))
```

```python
#Confusion matrix
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,model.predict(x_test))
print(cm)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,model.predict(x_test)))
```
